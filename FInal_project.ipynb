# Разработка модели машинного обучения для предсказания конечной температуры для металлургического комбината «Стальная птица»

<b> Описание проекта: </b> Чтобы оптимизировать производственные расходы, металлургический комбинат «Стальная птица» решил уменьшить потребление электроэнергии на этапе обработки стали. Для этого комбинату нужно контролировать температуру сплава. Заказчик планирует при помощи модели машинного обучения имитировать технологический процесс и предсказывать конечную температуру сплава из имющихся данных, что поможет ему добиться уменьшения энергозатрат. 

<b> Описание этапа обработки: </b>
Сталь обрабатывают в металлическом ковше вместимостью около 100 тонн. Чтобы ковш выдерживал высокие температуры, изнутри его облицовывают огнеупорным кирпичом. Расплавленную сталь заливают в ковш и подогревают до нужной температуры графитовыми электродами. Они установлены в крышке ковша. 

Из сплава выводится сера (этот процесс — десульфурация), добавлением примесей корректируется химический состав и отбираются пробы. Сталь легируют — изменяют её состав — подавая куски сплава из бункера для сыпучих материалов или проволоку через специальный трайб-аппарат (от англ. tribe — «масса»).

Перед тем как первый раз ввести легирующие добавки, измеряют температуру стали и производят её химический анализ. Потом температуру на несколько минут повышают, добавляют легирующие материалы и продувают сплав инертным газом. Затем его перемешивают и снова проводят измерения. Такой цикл повторяется до достижения целевого химического состава и оптимальной температуры плавки.

Тогда расплавленная сталь отправляется на доводку металла или поступает в машину непрерывной разливки. Оттуда готовый продукт выходит в виде заготовок-слябов (от англ. slab — «плита»).

<b> Цель исследования: </b> Построить модель машинного обучения, которая сможет предсказать конечную температуру сплава на основе предоставленной информации от заказчика.

<b> Ход исследования:</b> 

- Чтение датафреймов и их первичный осмотр: загрузка и изучение общей информации из представленных датасетов
- Исследовательский анализ данных: построение анализ качества предоставленных данных заказчиком, анализ признаков, выявление аномальных значений и выбросов
- Корреляционный анализ данных: поиск корреляций между данными, а также оценка мультиколлинеарности
- Объединение датафреймов: объединение датафреймов по ключам, отбор необходимых признаков для задачи, генерация новых признаков
- Исследовательский анализ данных объединённого датафрейма: построение анализ качества данных объединённого датафрейма, анализ признаков, выявление аномальных значений и выбросов
- Корреляционный анализ данных объединённого датафрейма: поиск корреляций между данными, а также оценка мультиколлинеарности
- Разработка модели машинного обучения: построение лучшей модели при помощи пайплайнов с различными видами моделей и гиперпараметров
- Проверка модели на тестовых данных и аналих её качества: проверка работы модели на тестовых данных, выявление важности входных признаков, а также проверка модели на адекватность 


<b> Общий вывод: </b> Выбор лучших моделей машинного обучения на основе метрики MAE, проверка выполнения требований заказчика, чтобы метрика MAE была ниже 6.8, а также рекомендации бизнесу для улучшения предсказаний по поддержанию температуры сплава.

При помощи данного исследования металлургический комбинат «Стальная птица» сможет предсказать конечную температуру сплава, что поможет уменьшить энергопотребление.

## Разработка модели машинного обучения

Целевым признаком задачи является температура сплава. Входными признаками являются объём материалов и объём проволочек (это важно, поскольку само сырьё, её объём и разновидность будет определять теплоёмкость полученного сплава), объём продувки газом (от этого будет зависеть степень остывания продукта между измерениями), средняя промежуточная температура и затраченная энергия на сплав. Перед нами стоит задача регрессии.

Разделим данные в пропорции 3:1 на тренировочную и тестовую.

```python
#Создадим переменные RANDOM_STATE и TEST_SIZE по требованию заказчика
RANDOM_STATE = 100225
TEST_SIZE = 0.25

X_train, X_test, y_train, y_test = train_test_split(
    df_full.drop('temperature', axis=1), 
    df_full['temperature'], 
    test_size = TEST_SIZE,
    random_state=RANDOM_STATE
)
```

```python
#Запомним наши столбцы
num_columns = list(X_train.columns)
```

```python
#Создаём общий пайплайн для подготовки данных
data_preprocessor = ColumnTransformer(
    [('num', MinMaxScaler(), num_columns)], 
    remainder='passthrough'
)
```

```python
#Создаём итоговый пайплайн: подготовка данных и модель
pipe_final = Pipeline([
    ('preprocessor', data_preprocessor),
    ('models', DecisionTreeRegressor())
 
])

param_grid = [
    #Словарь для модели LinearRegression()
    {
        'models': [LinearRegression()],
        'preprocessor__num': [StandardScaler(), MinMaxScaler()],    
    },

    #Словарь для модели Lasso()
    {
        'models': [Lasso()],
        'models__alpha': [0.001, 0.01, 0.1, 0.5, 1, 5, 10, 15, 20, 40],
        'preprocessor__num': [StandardScaler(), MinMaxScaler()],    
    },

    #Словарь для модели Ridge()
    {
        'models': [Ridge()],
        'models__alpha': [0.001, 0.01, 0.1, 0.5, 1, 5, 10, 15, 20, 40],
        'preprocessor__num': [StandardScaler(), MinMaxScaler()],    
    },
    
    
    # словарь для модели DecisionTreeRegressor()
    {
        'models': [DecisionTreeRegressor(random_state=RANDOM_STATE)],
        'models__max_depth': [1, 10, 100, 500],
        'models__min_samples_split': [5, 10, 100],
        'preprocessor__num': [StandardScaler(), MinMaxScaler()],
    }
    

]
```

```python
#Обучим модель и найдём наилучшую конфигурацию
model_Search = GridSearchCV(
    pipe_final, 
    param_grid, 
    cv=5,
    scoring='neg_mean_absolute_error',
    n_jobs=-1
)

model_Search.fit(X_train[num_columns], y_train);
    
print('Лучшая модель и её параметры:\n\n', model_Search.best_estimator_)
print ('Метрика лучшей модели на кросс-валидационной выборке:', round(-model_Search.best_score_, 2))
```

Модель машинного обучения была разработана и обучена при помощи GridSearchCV метода с пятью кросс-валидационными выборками, различными моделями и гиперпараметрами. Лучшей моделью является линейная модель регрессии Lasso(alpha=0.1), которая выдаёт на кросс-валидационной выборке метрику в 5.49, что ниже требуемой от заказчика метрики в 6.8.

```python
X_transform_test = model_Search.best_estimator_["preprocessor"].transform(X_test[num_columns])
predict = model_Search.best_estimator_["models"].predict(X_transform_test)
print(f'Метрика обученной модели на тестовой выборке: {round(mean_absolute_error(y_test, predict), 2)}')
```

```python
X_transform_train = model_Search.best_estimator_["preprocessor"].transform(X_train[num_columns])
model_dummy = DummyRegressor(strategy='mean')
model_dummy.fit(X_transform_train, y_train)
predict = model_dummy.predict(X_transform_test)
print(f'Метрика обученной dummy модели на тестовой выборке: {round(mean_absolute_error(y_test, predict), 2)}')
```

Как видно выше, метрика модели на тестовых данных равна 5.34, что ниже требуемой от заказчика метрики в 6.8. Также из проверки модели на адекватность при помощи среднего значения DummyRegressor видно, что наша обученная модель является правильной моделью машинного обучения, поскольку средние предсказанные значения будут давать метрику в 7.99.
